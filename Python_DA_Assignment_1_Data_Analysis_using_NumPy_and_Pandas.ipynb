###Numpy Array Operations:
To learn how to use NumPy for creating arrays, performing basic
array operations, and conducting numerical computations through analyzing
temperature data.

Analyzing the daily average temperatures (in degrees Celsius) data recorded over two weeks.

1. Create a 1D NumPy Array:

Created a 1D NumPy array named temperatures_w1 for Week 1

2. Inspection and Properties:

● Inspected the properties such as;

shape using ".shape",
data type using ".dtype",
the number of elements in the array using ".size" and
the no.of dimensions using ".ndim" of temperatures_w1 array.

3. Array Operations:

● Converted the temperatures_w1 from Celsius to Fahrenheit using the
formula: Fahrenheit = (Celsius * 9/5) + 32

● Finded the:
maximum temperature using ".max()", minimum temperature using ".min()", and mean temperatures for the week using ".mean()"(rounded off to 2 using round()).

4. Array Slicing and Indexing:

● Extracted the temperatures for the first three days of the week using ([:3]).

● Extracted the temperatures for the weekend (last two days) using ([-2:]).

● Extracted the temperatures for the middle three days of the week using ([2:5]).

5. Create a 2D Array:

Created a 2D NumPy array named temperatures where each row represents
temperatures for a different week:

★ Row 1: [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9] (Week 1)

★ Row 2: [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1] (Week 2)

6. Inspect and Slice the 2D Array:

● Inspected and printed
shape using ".shape",

data type  using ".dtype", and

total number of elements in the temperatures array using ".size" .

● Extract the temperatures for each week using slicing ([0]),([1])
and the weekends (last two days)for both weeks using slicing ([0][-2:]),([1][-2:]).

```python
import numpy as np
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
print("\nWeekly_average_temperatures:",temperatures_w1)
print("Shape:",temperatures_w1.shape)
print("Data_type:",temperatures_w1.dtype)
print("Number_of_elements:",temperatures_w1.size)
print("Number_of_dimensions:",temperatures_w1.ndim)

temp_farenheit = (temperatures_w1 * 9/5) + 32
print("\nWeekly_average_temperatures_in_farenheit:",temp_farenheit)

print("\nMaximum_temperature(*F):",temp_farenheit.max())
print("Minimum_temperature(*F):",temp_farenheit.min())
print("Mean_temperature(*F):",round(temp_farenheit.mean(),2))

print("\nTemperatures_for_first_three_days:",temp_farenheit[:3])
print("Temperatures_for_weekend:",temp_farenheit[-2:])
print("Temperatures_for_middle_three_days:",temp_farenheit[2:5])

temperatures=np.array([[22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],
                       [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]])
print("\nTemperatures_for_two_weeks:",temperatures)
print("Shape:",temperatures.shape)
print("Data_type:",temperatures.dtype)
print("Total_number_of_elements:",temperatures.size)

print("\nTemperatures_for_week1:",temperatures[0])
print("Temperatures_for_week2:",temperatures[1])

print("\nTemperatures_for_week1_weekend:",temperatures[0][-2:])
print("Temperatures_for_week2_weekend:",temperatures[1][-2:])

```

###Pandas Series:
1. Creating Pandas Series:

● Created a Pandas Series by importing panda [import pandas as pd] and named marks with the following data:

➢ Marks: 95, 92, 89, 85, 80

➢ Custom Indices: 'Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'

2. Indexing and Slicing:

● Used the integer index position to access the mark of the 1st rank student (marks[0]).

● Used the loc accessor to retrieve the marks of the top 3 ranks by
specifying their index labels (marks.loc['Rank1', 'Rank2', 'Rank3']).

● Used the iloc accessor to retrieve the mark of the 3rd rank student (marks.iloc[2]).

● Applied a boolean mask to filter the Series and retrieve the ranks where the marks are greater than 90 (marks[marks > 90]).

3. Manipulating Series:

● Modified the mark of the student with the 1st rank to 100 using (".loc").

● Removed the entry corresponding to the last rank from the Series using (".drop").

● Compute the CGPA by dividing each mark in the Series by 10 (marks/10).


```python
import pandas as pd
marks = pd.Series([95, 92, 89, 85, 80], index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'])
print(marks)

print("\nMark_of_1st_rank_student:",marks[0])
print("Marks_of_top_3_ranks:",marks.loc[['Rank1', 'Rank2', 'Rank3']])
print("Mark_of_3rd_rank_student:",marks.iloc[2])
print("Marks_greater_than_90:",marks[marks > 90])

marks.loc['Rank1'] = 100
print("\nUpdated_mark_of_1st_rank_student:",marks[0])

print(marks)

marks=marks.drop('Rank5')
print("\nUpdated_marks:",marks)

cgpa=marks/10
print("\nCGPA:",cgpa)

```

###Pandas DataFrame:

1. Creating Pandas DataFrame:

● Created a Pandas DataFrame by importing pandas [import pandas as pd] and named transactions with the following data:

➢ TransactionID: 101, 102, 103, 104, 105, 106, 107, 108, 109, 110

➢ ProductCategory: 'Electronics', 'Clothing', 'Electronics', 'Furniture',
'Clothing', 'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'

➢ Region: 'North', 'South', 'North', 'East', 'West', 'North', 'East', 'West','South', 'North'

➢ Amount: 200, 150, 300, 450, 200, 250, 300, 180, 350, 400

2. Data Exploration:

● Displayed the transactions DataFrame and its basic information, including head, tail, shape, column names, and data types using ".head(), .tail(), .shape(), .columns() and .dtypes()" respectively.

● Displayed only the 'ProductCategory' and 'Amount' columns.

● Retrieved the last 3 columns of the DataFrame using iloc ([:,-3:])

● Filtered rows where the 'Region' is 'North' and 'Amount' is greater than 200 (transactions['Region'] == 'North') & (transactions['Amount'] > 200).

● Displayed the value counts for the 'ProductCategory' column using (.value_counts()).

● Displayed the unique values in the 'Region' column using (.unique)

● Group by 'Region' and find the mean amount for each region using (.groupby()).

3. Manipulating the DataFrame:

● Modify the 'Amount' for the transaction with TransactionID 102 to 165 using (.loc())

● Added a new column 'Discount' by calculating 10% of the 'Amount'.

● Removed the row with TransactionID 109.

● Deleted the 'Discount' column from the DataFrame using (.drop())

```python
import pandas as pd

transactions = pd.DataFrame({
    "TransactionID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "product_category": ['Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing', 'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'],
    "Region":['North', 'South', 'North', 'East', 'West', 'North', 'East','West','South', 'North'],
    "Amount":[200, 150, 300, 450, 200, 250, 300, 180, 350, 400]})
print(transactions)
print("\nHead:",transactions.head())
print("Tail:",transactions.tail())
print("Shape:",transactions.shape)
print("Column_names:",transactions.columns)
print("Data_types:",transactions.dtypes)
print("\nProduct_category_v/s_Amount:",transactions[['product_category', 'Amount']])
print("\nLast_3_columns",transactions.iloc[:,-3:])
print("\nAmount>200_North",transactions[(transactions['Region'] == 'North') & (transactions['Amount'] > 200)])
print("\nProduct_count",transactions['product_category'].value_counts())
print("\nUnique_values_in_Region",transactions['Region'].unique())
print("\nRegion_Average_Amount",transactions.groupby('Region')['Amount'].mean())

transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165
print(transactions)

transactions['Discount'] = transactions['Amount'] * 0.10
print("\nNew_transactions_with_discounts",transactions)

transactions = transactions.drop(transactions[transactions['TransactionID'] == 109].index)
print("\nTransactions_after_deleting_109",transactions)

transactions = transactions.drop('Discount', axis=1)
print(transactions)




```
